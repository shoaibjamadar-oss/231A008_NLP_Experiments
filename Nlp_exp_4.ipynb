```python
print("Name: Shoaib Jamadar", "UIN: 231A008")
text_lower = text.lower()
print(text_lower)
```

```python
print("Name: Shoaib Jamadar", "UIN: 231A008")
from nltk.tokenize import word_tokenize
tokens = word_tokenize(text_lower)
print(tokens)
```

```python
print("Name: Shoaib Jamadar", "UIN: 231A008")
text_clean = []
for word in tokens:
    if word not in string.punctuation:
        text_clean.append(word)
print(text_clean)
```

```python
print("Name: Shoaib Jamadar", "UIN: 231A008")
# Step 3: Identify roots
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(tok, pos='v') for tok in text_clean]  # lemmatize as verbs
lemmas
```

```python
print("Name: Shoaib Jamadar", "UIN: 231A008")
lemmatizer.lemmatize("phone", pos='n')
```

```python
print("Name: Shoaib Jamadar", "UIN: 231A008")
[(token, lemma) for token, lemma in zip(tokens, lemmas)]
```

```python
print("Name: Shoaib Jamadar", "UIN: 231A008")
# Step 4: Analyze tokens manually for derivational vs inflectional
morphological_info = {}
for token, lemma in zip(tokens, lemmas):
    if token != lemma:
        if token.endswith("ing"):
            analysis = "Inflectional: Present participle/continuous form of '{}'".format(lemma)
        elif token.endswith("er") or token.endswith("ation"):
            analysis = "Derivational: '{}' + -er → person who does the action".format(lemma)
        else:
            analysis = "Inflectional or irregular form of '{}'".format(lemma)
    else:
        analysis = "Base/root form"
    morphological_info[token] = analysis
```

```python
print("Name: Shoaib Jamadar", "UIN: 231A008")
morphological_info
```

```python
print("Name: Shoaib Jamadar", "UIN: 231A008\n")
# Step 5: Generate word forms manually
word_forms = {
    "run": ["running", "ran", "runner"]
}

# Step 6: Summarize findings
print("Original text:", text)
print("\nTokens:", tokens)
print("\nLemmas:", lemmas)
print("\nMorphological Analysis:")
for word, analysis in morphological_info.items():
    print(f"  {word}: {analysis}")

print("\nGenerated word forms from root 'run':", word_forms["run"])
```
