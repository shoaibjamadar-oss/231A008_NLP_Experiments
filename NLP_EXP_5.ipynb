```python
print("Name: Shoaib jamadar","UIN:231A008")
import nltk
from nltk import word_tokenize, bigrams, FreqDist, ConditionalFreqDist
from nltk.corpus import stopwords
from collections import defaultdict

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
#Step 1: Load and preprocess the text corpus
def preprocess(text):
    #Convert to Lowercase
    text = text.lower()
    #Tokenize the text
    tokens = word_tokenize(text)
    #Remove punctuation and stopwords
    stop_words = set(stopwords.words('english'))
    tokens = [token for token in tokens if token.isalnum() and token not in stop_words]
    return tokens

#Sample text corpus
corpus = "The quick brown fox jumps over the lazy dog. The fox is quick and the dog is lezy. The quick brown fox could not jump."
#Step 2: Preprocess the text
tokens = preprocess(corpus)
print(tokens)
```

```python
print("Name: Shoaib jamadar","UIN:231A008")
from nltk.util import ngrams

#Step 3: Generate bigrams from the tokenized text
bigrams_lists = list(bigrams(tokens))
print("Bigrams:", bigrams_lists)

# Generate trigrams from the tokenized text
trigrams_lists = list(ngrams(tokens, 3))
print("Trigrams:", trigrams_lists)
```

```python
print("Name: Shoaib jamadar","UIN:231A008")
#Step 4: Count bigram frequencies using FreqDist
bigram_freq = FreqDist(bigrams_lists)
bigram_freq
```

```python
print("Name: Shoaib jamadar","UIN:231A008")
#Step 5: Calculate conditional frequencies to predict the next word
cfd = ConditionalFreqDist(bigrams_lists)
print("Conditional Frequencies:", cfd.keys())
```

```python
cfd.items()
```

```python
trigram_freq = FreqDist(trigrams_lists)
trigram_freq
```

```python
# Step 6: Function to predict the next word based on a given word
def predict_next_word(word):
    word = word.lower()
    if word in cfd:
        next_word = cfd[word].max() # Get the most likely word following the given word
        return next_word
    else:
        return None
```

```python
#Test the prediction function
start_word="fox"
predicted_word = predict_next_word(start_word)
if predicted_word:
    print(f"Next word after '{start_word}': {predicted_word}")
else:
    print(f"No prediction available for the word {start_word}.")
```

```python
#Step 7: Generate a sequence of words
def generate_sequence(start_word, length):
    sequence = [start_word]
    current_word = start_word
    for _ in range(length - 1):
        next_word = predict_next_word(current_word)
        if next_word:
            sequence.append(next_word)
            current_word = next_word
        else:
            break
    return ' '.join(sequence)

#Generate a sequence starting from 'fox'
generated_sequence = generate_sequence("fox", length=10)
print(f"Generated sequence: {generated_sequence}")
```
