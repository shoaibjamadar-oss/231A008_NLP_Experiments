```python
text='''Hey everyone!!! 👋

I'm REALLY excited 😊 because I've just completed my first Machine Learning project on 23rd July, 2026. The project analyzed over 12,500 customer reviews collected from https://www.example.com and another dataset downloaded from www.kaggle.com.

Unfortunately, some reviews weren't useful—they contained only words like "the", "is", "and", "of", "to", etc. Others had strange symbols such as @#$%^&*, repeated punctuation!!!!!!, HTML tags like <br>, and emojis 😂😂😂.

You can contact me at vijay.shanker@eng.rizvi.edu.in or support123@company.co.in if you'd like the dataset. Follow us on Twitter @AI_ResearchLab and use hashtags like #MachineLearning #NLP #DataScience.

The model's accuracy increased from 82.5% to 91.8%, while the training time decreased by nearly 35%. It processed 1,25,000 records in less than 2.5 hours and consumed around 3.7 GB of RAM.

Students were studying, analyzing, processing, predicting, running, writing, and comparing different algorithms. The researchers discovered better techniques after performing extensive experiments and evaluating multiple models.

Don't worry if the results aren't perfect—you're learning! Let's improve them together. I'll upload the final report by 30/07/2026.

Thanks!!!'''
```

```python
print("Shoaib Jamadar","UIN: 231A008")
small_text=text.lower()
print(small_text)
```

```python
print("Shoaib Jamadar", "UIN: 231A008\n")
import re
#by executing this removing tweeter handles,Emails,url
pattern = r'(?:https?://\S+|www\.\S+|\S+@\S+|@\S+)'
text_url = re.sub(pattern, " ", small_text)
print(text_url)
```

```python
!pip install emoji
import emoji
text_emoji = emoji.replace_emoji(text_url, replace="")
print(text_emoji)
```

```python
text_no_numbers = re.sub(r'\d+|%', ' ', text_emoji)
print(text_no_numbers)
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize

# Tokenize the text
tokens = word_tokenize(text_no_numbers)

# Print the first 20 tokens to inspect
print(tokens[:20])
```

```python
print("Shoaib Jamadar", "UIN: 231A008\n")
import string
punctuation=list(string.punctuation)
print(punctuation)
```

```python
print("Shoaib Jamadar", "UIN: 231A008\n")
print("With punctuation:", tokens)
print()
text_punct=[]
for word in tokens:
    if word not in punctuation:
        text_punct.append(word)
print("Without punctuation:", text_punct)
```

```python
from nltk.corpus import stopwords
import nltk

# Download the stopwords corpus
try:
    stopwords.words('english')
except LookupError:
    nltk.download('stopwords')

# Get English stopwords
stop_words = set(stopwords.words('english'))

# Remove stopwords from text_punct
text_no_stopwords = [word for word in text_punct if word.lower() not in stop_words]

print("Without stopwords:", text_no_stopwords)
```

```python
from nltk.stem import PorterStemmer
import nltk


# Initialize Porter Stemmer
porter = PorterStemmer()

# Perform stemming
text_stemmed = [porter.stem(word) for word in text_no_stopwords]

print("Stemmed text:", text_stemmed)
```

```python
from nltk.stem import WordNetLemmatizer
import nltk

lemmatizer = WordNetLemmatizer()

text_lemmatized = [lemmatizer.lemmatize(word) for word in text_no_stopwords]

print("Lemmatized text:", text_lemmatized)
```
