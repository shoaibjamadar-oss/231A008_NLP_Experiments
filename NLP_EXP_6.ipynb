```python
print("Name: Shoaib jamadar", "UIN:231A008")

import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import UnigramTagger, BigramTagger, pos_tag
from nltk.corpus import treebank
from nltk import pos_tag_sents

# Download required NLTK data files
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')  # needed on newer NLTK versions
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('treebank')
```

```python
print("Name: Shoaib jamadar", "UIN:231A008")

# Sample text for tagging
sample_text = "He is reading a book. Book a flight ticket for me."

# Tokenize the text
tokens = word_tokenize(sample_text)
print(tokens)
```

```python
print("Name: Shoaib jamadar", "UIN:231A008")

# Rule-based POS Tagging using NLTK's built-in tagger (Perceptron-based, statistical)
rule_based_tags = pos_tag(tokens)
print(rule_based_tags)
```

```python
print("Name: Shoaib jamadar", "UIN:231A008")

# Look up what a POS tag means
nltk.help.upenn_tagset('VB')
```

```python
print("Name: Shoaib jamadar", "UIN:231A008")

nltk.help.upenn_tagset('VBZ')
```

```python
print("Name: Shoaib jamadar", "UIN:231A008")

# Statistical Tagging using Treebank corpus-trained taggers
train_sents = treebank.tagged_sents()

# Unigram Tagger
unigram_tagger = UnigramTagger(train_sents)
unigram_tags = unigram_tagger.tag(tokens)
print(unigram_tags)
```

```python
print("Name: Shoaib jamadar", "UIN:231A008")

# Bigram Tagger (falls back to the Unigram Tagger when a bigram is unseen)
bigram_tagger = BigramTagger(train_sents, backoff=unigram_tagger)
bigram_tags = bigram_tagger.tag(tokens)
print(bigram_tags)
```

```python
print("Name: Shoaib jamadar", "UIN:231A008")

# Number of tagged sentences used to train the taggers
len(train_sents)
```

```python
print("Name: Shoaib jamadar", "UIN:231A008")

# Display the results for comparison
print("Original Sentence:", sample_text)
print("\nRule-Based POS Tagging (Statistical):")
print(rule_based_tags)
print("\nUnigram POS Tagging:")
print(unigram_tags)
print("\nBigram POS Tagging:")
print(bigram_tags)
```

```python
print("Name: Shoaib jamadar", "UIN:231A008")

import pandas as pd

# Side-by-side comparison of Rule-Based, Unigram and Bigram tags, token by token
comparison_df = pd.DataFrame({
    "Token": [w for w, t in rule_based_tags],
    "Rule-Based Tag": [t for w, t in rule_based_tags],
    "Unigram Tag": [t for w, t in unigram_tags],
    "Bigram Tag": [t for w, t in bigram_tags]
})

comparison_df
```

```python
print("Name: Shoaib jamadar", "UIN:231A008")

# Evaluate the performance of each tagger using held-out Treebank sentences
test_sents = treebank.tagged_sents()[1000:]

# .accuracy() is the modern method name; .evaluate() is kept as a fallback for older NLTK versions
unigram_acc = unigram_tagger.accuracy(test_sents) if hasattr(unigram_tagger, "accuracy") else unigram_tagger.evaluate(test_sents)
bigram_acc = bigram_tagger.accuracy(test_sents) if hasattr(bigram_tagger, "accuracy") else bigram_tagger.evaluate(test_sents)

print(f"\nUnigram Tagger Accuracy: {unigram_acc * 100:.2f}%")
print(f"Bigram Tagger Accuracy: {bigram_acc * 100:.2f}%")
```

```python
print("Name: Shoaib jamadar", "UIN:231A008")

import matplotlib.pyplot as plt

# Side-by-side accuracy comparison chart
taggers = ["Unigram Tagger", "Bigram Tagger"]
accuracies = [unigram_acc * 100, bigram_acc * 100]

plt.figure(figsize=(6, 4))
plt.bar(taggers, accuracies, color=["skyblue", "salmon"])
plt.ylabel("Accuracy (%)")
plt.title("Unigram vs Bigram Tagger Accuracy")
plt.ylim(0, 100)
for i, acc in enumerate(accuracies):
    plt.text(i, acc + 1, f"{acc:.2f}%", ha="center")
plt.show()
```
